```python
import random
import os

def randomize_and_split_train(file_paths, output_dir):
    # Randomize the list
    random.shuffle(file_paths)

    # Determine the size of each split
    split_size = len(file_paths) // 4

    # Split the list into 4 parts
    splits = [file_paths[i*split_size:(i+1)*split_size] for i in range(4)]

    # Ensure the last list contains any remaining elements
    if len(file_paths) % 4 != 0:
        splits[-1].extend(file_paths[4*split_size:])

    # Save each split to a text file
    for i, split in enumerate(splits):
        with open(os.path.join(output_dir, f'file_paths_train-{i+1}.txt'), 'w') as f:
            for path in split:
                f.write(f"{path}\n")

# Example usage
output_dir = '/content/drive/MyDrive/Featurenet_exp/names'
randomize_and_split_train(file_paths_train, '/content/drive/MyDrive/Featurenet_exp/names')
```

```python
import os
file_paths_train_1=[]
directory = '/content/drive/MyDrive/Featurenet_exp/names'

with open(os.path.join(directory, 'file_paths_train-1.txt'), 'r') as f:
    file_paths_train_1 = [line.strip() for line in f]
```

```python
len(file_paths_train_1)
```

```python
def extract_labels (path):
  with open(path, 'rb') as fp:

    file_name=os.path.basename(fp.name)
    label=file_name.split('_')[0]
    label=int(label)
    return label
```

```python
labels_train_1=[extract_labels(path) for path in file_paths_train_1]
labels_train_1[:10]
```

```python
if len(labels_train_1)==len(file_paths_train_1):
  print('Go ahead things are working well')
else:
  print('check file directory')
```

```python
unique_labels=np.unique(labels_train_1)
unique_labels.shape
```

```python
def get_label(file_paths):
  labels=[]
  for path in file_paths:
    label=extract_labels(path)
    labels.append(label)
  return labels
```

```python
label_check_1=get_label(file_paths_train_1)
```
